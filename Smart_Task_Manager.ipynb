```python
# Step 0: Install libraries
!pip install pdfplumber nltk pandas openpyxl --quiet

# Step 1: Import libraries
import pdfplumber
import nltk
from nltk.tokenize import sent_tokenize
import pandas as pd

# Step 2: Download NLTK data
nltk.download('punkt')

# Step 3: Upload PDF in Colab
from google.colab import files
uploaded = files.upload()

pdf_path = list(uploaded.keys())[0]  # Gets the uploaded file name

# Step 4: Extract text from PDF
text = ""
try:
    with pdfplumber.open(pdf_path) as pdf:
        for page in pdf.pages:
            page_text = page.extract_text()
            if page_text:  # Avoid NoneType
                text += page_text + "\n"
except Exception as e:
    print("Error reading PDF:", e)

print("Preview of extracted text:\n", text[:500])

# Step 5: Split text into sentences
sentences = sent_tokenize(text)

# Step 6: Extract engineering tasks based on keywords
keywords = ["issue", "problem", "check", "verify", "review", "test", "update"]
tasks = [sent for sent in sentences if any(word in sent.lower() for word in keywords)]

# Step 7: Create checklist
checklist = pd.DataFrame(tasks, columns=["Task Description"])
checklist.index += 1  # Start numbering from 1
print("\nGenerated Checklist:\n", checklist)

# Step 8: Save checklist to Excel
checklist_file = "/content/engineer_checklist.xlsx"
checklist.to_excel(checklist_file, index_label="No")
print(f"\nChecklist saved as '{checklist_file}'")

# Step 9: Generate simple communication summary
summary_sentences = sentences[:5]  # first 5 sentences as example
summary = " ".join(summary_sentences)
print("\nCommunication Summary:\n", summary)
```
